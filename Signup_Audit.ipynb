Standardized plan_selected and gender to fix casing and typos, using a dictionary of known variants mapped to clean categories.

```python
plan_map = {
    'basic': 'Basic',
    'basci': 'Basic',
    'pro': 'Pro',
    'proe': 'Pro',
    'premium': 'Premium',
    'premum': 'Premium',
}
```

```python
def clean_plan(val):
    key = str(val).strip().lower()
    return plan_map.get(key, str(val).strip().title())
```

```python
df['plan_selected'] = df['plan_selected'].apply(clean_plan)
```

```python
df['plan_selected'].unique()
```

```python
gender_map = {
    'male': 'Male',
    'man': 'Male',
    'm': 'Male',
    'female': 'Female',
    'woman': 'Female',
    'f': 'Female',
    'nb': 'Other',
    'non-binary': 'Other',
}
```

```python
def clean_gender(val):
    if pd.isna(val):
        return np.nan
    key = str(val).strip().lower()
    return gender_map.get(key, str(val).strip().title())
```

Missing or malformed emails (26 rows) flagged as invalid rather than guessed — these customers are unreachable by email until corrected at the source.

```python
import re

email_pattern = re.compile(r'^[^@\s]+@[^@\s]+\.[^@\s]+$')
```

```python
def is_valid_email(val):
    if pd.isna(val):
        return False
    return bool(email_pattern.match(str(val)))
```

```python
df['email_valid'] = df['email'].apply(is_valid_email)
```

```python
df[df['email_valid'] == False]
```

```python
df.loc[df['email_valid'] == False, 'email'] = np.nan
```
